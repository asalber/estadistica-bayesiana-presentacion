## Ejercicio 9 — Simulación de la distribución a posteriori Gamma-Poisson

Partiendo de la distribución a posteriori anterior para la tasa de infartos, $\lambda \sim \text{Gamma}(55, 7)$, simula la distribución a posteriori si se toma otra muestra de tamaño 5 en la que se han producido $X_1=5, X_2=7, X_3=4, X_4=6, X_5=3$.

```r
# Rejilla de posibles valores de λ
rejilla <- seq(1, 15, length.out = 100)
# Función de verosimilitud para la nueva muestra
verosimilitud <- sapply(rejilla, function(l) prod(dpois(c(...), lambda = l)))
# Distribución a priori (a posteriori anterior)
prior <- dgamma(rejilla, shape = ..., rate = ...)
# Plausibilidad a posteriori
posterior <- prior * verosimilitud
# Normalización
df_posterior <- tibble(λ = rejilla, posterior = ... / ...)
set.seed(123)
# Muestra aleatoria de la distribución a posteriori
muestra_posterior <- slice_sample(df_posterior, n = 10000, weight_by = posterior, replace = TRUE)
# Histograma con la distribución teórica superpuesta
ggplot(muestra_posterior, aes(x = λ)) +
  geom_histogram(aes(y = after_stat(density)), binwidth = 0.5, alpha = 0.7, fill = "steelblue", color = "white") +
  stat_function(fun = dgamma, args = list(..., ...), color = "red") +
  labs(title = "Distribución a posteriori (simulación de rejilla)",
       x = "Tasa de infartos (λ)", y = "Densidad") +
  theme_minimal()
```
